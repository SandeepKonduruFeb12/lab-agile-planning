Knn classification of the employee attrition data


Step1: import the required modules

```
from pandas import read_csv
from sklearn.model_selection import KFold
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn import preprocessing
from sklearn.neighbors import KNeighborsClassifier

```

# step 2: read the data from the csv file and split it to input and output


```
filename = 'https://raw.githubusercontent.com/sandeepkhcltech/aiml/refs/heads/main/emp_attr_knn_data.csv'
names = ['Age','JobRole','MonthlyIncome','JobSatisfaction','YearsAtCompany','Attrition']
df = read_csv(filename, names=names, skiprows=1) # Skip the header row
print (df)
```

Step 3: split to input and output



```
# Encode the 'JobRole' column
le = preprocessing.LabelEncoder()
df['JobRole'] = le.fit_transform(df['JobRole'])

array = df.values
inputx = array[:,0:4]
outputy = array[:,5] # Select the 'Attrition' column as output
print (inputx)
print (outputy)
```

Step 4: Select the model

```
model = KNeighborsClassifier()
print("\nThe model selected is",model)
```

# step 6: train or build the model


```
model.fit(inputx,outputy)
```

# step 7: do testing or model prediction

```
filename = 'https://raw.githubusercontent.com/sandeepkhcltech/aiml/refs/heads/main/emp_attr_knn_test.csv'
names = ['Age','JobRole','MonthlyIncome','JobSatisfaction','YearsAtCompany']
newdataframe = read_csv(filename, names=names, skiprows=1)

# Encode the 'JobRole' column in the test data using the same encoder
newdataframe['JobRole'] = le.transform(newdataframe['JobRole'])

array = newdataframe.values
z = array[:,0:4]
print("\n",newdataframe,"\n")
res=model.predict(z)
```

# Step 8: Print the predicted results

```
# step 8: visualize the test results
print("\nThe predicted results are:")
print(res)
```
